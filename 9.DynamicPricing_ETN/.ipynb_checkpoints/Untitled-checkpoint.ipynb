```python
def load_and_aggregate(D4NN, target_col):
    """
    Agrupa por FECHA_CORRIDA y conserva columnas originales (sin PORCENT_PROMO).
    """
    Frame = D4NN[['FECHA_OPERACION', target_col, 'MES', 'FOP']].copy()
    agg_dict = {col: 'mean' for col in Frame.columns if col not in ['FECHA_OPERACION', target_col]}
    agg_dict[target_col] = 'sum'

    df = Frame.groupby('FECHA_OPERACION').agg(agg_dict)
    df.index = pd.to_datetime(df.index)
    return df.sort_index()

col= 'VENTA' 
df= load_and_aggregate(D4NN, col)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.fft import rfft, rfftfreq, irfft

# Suponiendo que 'df' y la columna 'VENTA' ya están cargados.
# Ejemplo de datos (sustituye esto por tu carga real de df):
# df = pd.DataFrame({'VENTA': np.sin(np.linspace(0, 10*np.pi, 365)) + 0.5 * np.random.randn(365)})
# df['FECHA'] = pd.to_datetime(pd.date_range(start='2024-01-01', periods=365))
# df = df.set_index('FECHA')

# --- Preparación de datos ---
# Asegúrate de que df tenga índice datetime y una columna "VENTA"
series = df['VENTA'].values
n = len(series)
dt = 1  # datos diarios (o el intervalo de tiempo entre tus datos)

# 1. --- Transformada Rápida de Fourier (FFT) ---
fft_vals = rfft(series)
fft_freqs = rfftfreq(n, dt)

# 2. --- Aplicar la Derivada en el Dominio de la Frecuencia ---
# La derivada en el dominio del tiempo (d/dt) es equivalente a:
# F'(k) = F(k) * (i * 2 * pi * k)
# Donde F(k) son los valores de la FFT y k son las frecuencias.

# Asegúrate de que la frecuencia cero sea manejada correctamente.
# rfft_freqs incluye solo frecuencias no negativas.

# Multiplicador para la derivada
# i es la unidad imaginaria (1j en Python)
derivative_multiplier = 1j * 2 * np.pi * fft_freqs

# Derivada de los valores de FFT (que ahora incluyen parte real e imaginaria)
fft_derivative_vals = fft_vals * derivative_multiplier

# 3. --- Análisis del Espectro de la Derivada ---

# Obtener Amplitud del espectro de la derivada
derivative_amplitudes = np.abs(fft_derivative_vals)

# Ignorar frecuencia cero (k=0) en el análisis de amplitud
# La frecuencia cero siempre tiene una amplitud de 0 después de la derivada
# ya que derivative_multiplier[0] es 0.
fft_freqs_clean = fft_freqs[1:]
derivative_amplitudes_clean = derivative_amplitudes[1:]

# Periodo dominante de la DERIVADA
# Esto te mostrará el periodo que tiene la mayor TASA de CAMBIO
dominant_period_derivative = 1 / fft_freqs_clean[np.argmax(derivative_amplitudes_clean)]
print("Periodo dominante de la DERIVADA estimado:", dominant_period_derivative, "días")

# 4. --- Graficar Espectro de la Derivada ---
plt.figure(figsize=(12,5))
plt.plot(1/fft_freqs_clean, derivative_amplitudes_clean)
plt.xlabel("Periodo (días)")
plt.ylabel("Amplitud del Espectro de la Derivada")
plt.title("Espectro de Potencia de la Serie Derivada (FFT)")
#plt.xscale('log') # Escala logarítmica para ver mejor los periodos largos
plt.xlim(2,20)
plt.grid(True, which="both", ls="--")
plt.show()

# 5. --- (Opcional) Reconstruir la Serie Derivada en el Tiempo ---
# Para ver la serie derivada en el dominio del tiempo
derived_series = irfft(fft_derivative_vals, n)

plt.figure(figsize=(12,5))
plt.plot(df.index, series, label='Serie Original (VENTA)')
plt.plot(df.index, derived_series, label='Serie Derivada (d(VENTA)/dt)', linestyle='--')
plt.title("Comparación: Serie Original vs. Derivada (Calculada por FFT)")
plt.xlabel("Fecha")
plt.ylabel("Valor / Tasa de Cambio")
plt.legend()
plt.grid(True)
plt.show()
```

```python
"""
seasonality_diagnostics.py

Módulo para detectar periodo real, armónicos y tipo de estacionalidad
según la lógica solicitada:

1) FFT de la serie (0..max_period_plot)
2) FFT de la derivada
3) ACF/PACF de la serie y derivada
4) Clasificación candidato (real / armónico / ruido)
5) Confirmación por maximización de var(seasonal)/var(y) con STL
6) Determinar si la serie es multiplicativa o aditiva

Funciones principales:
- analyze_series(series, ...)

Devuelve un dict con resultados y guarda/retorna figuras.

Requisitos: numpy, pandas, scipy, statsmodels, matplotlib
"""

from typing import List, Tuple, Dict, Any, Optional
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.fft import rfft, rfftfreq
from scipy.signal import find_peaks
from statsmodels.tsa.stattools import acf, pacf
from statsmodels.tsa.seasonal import STL
from scipy.stats import boxcox
import warnings

# ---------------------------
# Helpers / defaults
# ---------------------------

DEFAULTS = {
    "dt": 1.0,  # days
    "max_period_plot": 20,
    "nlags_acf": 60,
    "fft_energy_thresh": 0.10,  # relative to max
    "tolerance_abs": 0.5,  # days
    "tolerance_rel": 0.07,  # relative
    "window_len_trend": 31,
}


def ensure_daily_series(series: pd.Series) -> pd.Series:
    """Return series reindexed at daily frequency, forward-filled.
    Assumes series index is datetime-like.
    """
    if not isinstance(series.index, pd.DatetimeIndex):
        raise ValueError("series must have a DatetimeIndex")
    s = series.copy().sort_index()
    s = s.asfreq('D')
    # forward fill then backfill if leading NaNs
    s = s.fillna(method='ffill').fillna(method='bfill')
    return s


def detrend_series(series: pd.Series, window: int = None) -> Tuple[pd.Series, pd.Series]:
    """Compute a robust trend (rolling median) and detrended series.
    Returns (trend, detrended)
    """
    if window is None:
        window = DEFAULTS['window_len_trend']
    trend = series.rolling(window, center=True, min_periods=max(3, window//3)).median()
    # fallback to rolling mean if many NaNs
    trend = trend.fillna(series.rolling(window, center=True, min_periods=1).mean())
    detrended = series - trend
    detrended = detrended.fillna(method='ffill').fillna(method='bfill')
    return trend, detrended


def compute_fft(series: pd.Series, dt: float = 1.0, windowed: bool = True,
                max_period_plot: float = 20.0) -> Dict[str, np.ndarray]:
    """Compute rFFT and return periods (in days), amplitudes and frequencies.
    Filters out zero-frequency and limits to periods <= max_period_plot.
    """
    x = series.values.astype(float)
    n = len(x)
    if n < 4:
        raise ValueError("series too short for FFT")
    if windowed:
        w = np.hanning(n)
        x = x * w
    fft_vals = rfft(x)
    freqs = rfftfreq(n, dt)
    # remove zero freq
    mask = freqs > 0
    freqs = freqs[mask]
    amps = np.abs(fft_vals)[mask]
    # convert to periods
    periods = 1.0 / freqs
    # sort by period ascending
    order = np.argsort(periods)
    periods = periods[order]
    amps = amps[order]
    # limit to max_period_plot
    mask2 = periods <= max_period_plot
    return {
        'periods': periods[mask2],
        'amps': amps[mask2],
        'freqs': freqs[order][mask2]
    }


def find_fft_peaks(periods: np.ndarray, amps: np.ndarray,
                   energy_thresh: float = 0.10) -> List[Tuple[float, float]]:
    """Return list of (period, amplitude) peaks above relative threshold.
    Uses scipy.find_peaks on amplitude series (on the given ordering by period).
    """
    if len(periods) == 0:
        return []
    # simple peak-finding on amplitude curve
    peaks_idx, _ = find_peaks(amps)
    if peaks_idx.size == 0:
        return []
    peak_periods = periods[peaks_idx]
    peak_amps = amps[peaks_idx]
    # threshold relative to global max
    thresh = energy_thresh * amps.max()
    candidates = [(float(p), float(a)) for p, a in zip(peak_periods, peak_amps) if a >= thresh]
    # sort descending by amplitude
    candidates = sorted(candidates, key=lambda x: -x[1])
    return candidates


def compute_acf_pacf(series: pd.Series, nlags: int = 60) -> Dict[str, np.ndarray]:
    """Compute ACF and PACF values.
    Returns dict with 'acf' and 'pacf' arrays (including lag 0).
    """
    x = series.dropna().values
    if len(x) < 3:
        raise ValueError("series too short for ACF/PACF")
    acf_vals = acf(x, nlags=nlags, fft=True)
    pacf_vals = pacf(x, nlags=nlags, method='yw')
    return {'acf': acf_vals, 'pacf': pacf_vals}


def significant_acf_lags(acf_vals: np.ndarray, n: int) -> List[int]:
    """Return lags where acf is statistically significant at approx 95%.
    Uses approximate conf interval +/-1.96/sqrt(n)
    """
    conf = 1.96 / np.sqrt(n)
    lags = list(np.where(np.abs(acf_vals) > conf)[0])
    # remove lag 0 from list unless desired
    if 0 in lags:
        lags.remove(0)
    return lags


def is_harmonic(candidate_p: float, other_ps: List[float], tol_abs: float = 0.5,
                tol_rel: float = 0.07, max_k: int = 6) -> Optional[Tuple[float,int]]:
    """Check if candidate_p is harmonic of another period in other_ps.
    Returns (base_period, k) if harmonic found, else None.
    harmonic means candidate ≈ base/k or candidate ≈ base*k for integer k.
    """
    for p0 in other_ps:
        if p0 <= 0:
            continue
        for k in range(2, max_k+1):
            # candidate approx p0 / k
            if abs(candidate_p - p0 / k) <= max(tol_abs, tol_rel * p0):
                return (p0, k)
            # candidate approx p0 * k
            if abs(candidate_p - p0 * k) <= max(tol_abs, tol_rel * p0):
                return (p0, k)
    return None


def classify_fft_candidates(candidates_fft: List[Tuple[float, float]],
                            candidates_fft_diff: List[Tuple[float, float]],
                            acf_lags: List[int], acf_lags_diff: List[int],
                            tol_abs: float = 0.5, tol_rel: float = 0.07,
                            energy_thresh: float = 0.10) -> List[Dict[str, Any]]:
    """Classify each candidate period as 'real', 'armonic', or 'noise'.
    Combines info from FFT on series and derivative and ACFs.
    Returns list of dicts with classification info.
    """
    # gather all FFT period candidates
    all_candidates = []
    # unique set of period values (rounded a bit) preserve amplitude from main list
    seen = set()
    for p, a in candidates_fft + candidates_fft_diff:
        key = round(p, 3)
        if key not in seen:
            seen.add(key)
            all_candidates.append({'period': float(p), 'amplitude': float(a)})

    # list of period floats for harmonic tests
    period_list = [c['period'] for c in all_candidates]

    results = []
    for c in all_candidates:
        p = c['period']
        a = c['amplitude']
        # ACF match? check if any acf_lag ~= p within tolerance
        acf_match = any(abs(l - p) <= max(tol_abs, tol_rel * p) for l in acf_lags)
        acf_diff_match = any(abs(l - p) <= max(tol_abs, tol_rel * p) for l in acf_lags_diff)
        # harmonic test against other periods
        harmonic = is_harmonic(p, [pp for pp in period_list if pp != p], tol_abs=tol_abs, tol_rel=tol_rel)
        # energy test
        is_noise = (a < energy_thresh * max([xx['amplitude'] for xx in all_candidates])) and (not acf_match)
        # classification rules
        if acf_match or acf_diff_match:
            if harmonic and not acf_match:
                classification = 'armonic'  # supported only in derivative
            else:
                classification = 'real'
        else:
            if harmonic:
                classification = 'armonic'
            elif is_noise:
                classification = 'noise'
            else:
                classification = 'uncertain'
        results.append({
            'period': p,
            'amplitude': a,
            'acf_match': acf_match,
            'acf_diff_match': acf_diff_match,
            'harmonic_of': harmonic,
            'is_noise': is_noise,
            'classification': classification
        })
    # sort by amplitude descending
    results = sorted(results, key=lambda x: -x['amplitude'])
    return results


def stl_score_for_period(series: pd.Series, period: int) -> float:
    """Run STL and return var(seasonal)/var(y) score.
    period must be integer > 1
    """
    if period < 2:
        return 0.0
    # STL needs no NaNs
    y = series.dropna()
    if len(y) < max(10, 3*period):
        # warn that estimation unreliable
        warnings.warn(f"Short series for robust STL at period={period}")
    res = STL(y, period=period, robust=True).fit()
    score = float(np.var(res.seasonal) / np.var(y)) if np.var(y) > 0 else 0.0
    return score


def choose_best_period_by_stl(series: pd.Series, candidate_periods: List[float]) -> Tuple[int, List[Tuple[int, float]]]:
    """From candidate periods (floats), run STL on rounded ints and pick best by var ratio.
    Returns (best_period_int, list_of_(period_int,score) sorted desc).
    """
    cand_ints = sorted(set(max(2, int(round(p))) for p in candidate_periods))
    scores = []
    for p in cand_ints:
        score = stl_score_for_period(series, p)
        scores.append((p, score))
    scores_sorted = sorted(scores, key=lambda x: -x[1])
    best = scores_sorted[0][0] if scores_sorted else 0
    return best, scores_sorted


def determine_additive_or_multiplicative(series: pd.Series, best_period: int) -> Tuple[str, Dict[str, Any]]:
    """Decide if series is multiplicative or additive.
    Returns ('multiplicative'|'additive', diagnostics dict)
    """
    y = series.dropna()
    if best_period < 2:
        return 'additive', {'reason': 'no seasonality detected'}
    res = STL(y, period=best_period, robust=True).fit()
    trend = res.trend
    seasonal = res.seasonal
    # scatter correlation between abs(seasonal) and trend
    corr = np.corrcoef(np.nan_to_num(trend), np.abs(np.nan_to_num(seasonal)))[0,1]
    # boxcox lambda
    eps = max(1e-6, np.percentile(y, 1) * 0.001) if np.min(y) <= 0 else 0
    y_pos = y + eps
    try:
        y_bc, lam = boxcox(y_pos)
    except Exception:
        lam = np.nan
    diagnostics = {'corr_trend_season_abs': float(corr), 'boxcox_lambda': float(lam) if not np.isnan(lam) else None}
    # rules
    if (not np.isnan(lam) and lam < 0.3) or (corr > 0.35):
        return 'multiplicative', diagnostics
    else:
        return 'additive', diagnostics


# ---------------------------
# Main analysis function
# ---------------------------

def analyze_series(series: pd.Series,
                   col_name: Optional[str] = None,
                   dt: float = DEFAULTS['dt'],
                   max_period_plot: float = DEFAULTS['max_period_plot'],
                   nlags_acf: int = DEFAULTS['nlags_acf'],
                   fft_energy_thresh: float = DEFAULTS['fft_energy_thresh'],
                   tolerance_abs: float = DEFAULTS['tolerance_abs'],
                   tolerance_rel: float = DEFAULTS['tolerance_rel']
                   ) -> Dict[str, Any]:
    """Run the full pipeline and return a diagnostics dictionary.

    series : pd.Series with DatetimeIndex
    Returns dict with keys: fft, fft_diff, acf/pacf, classified_candidates, stl_scores, best_period,
    seasonality_type, figures (matplotlib Figure objects)
    """
    # 1. prepare
    sname = col_name or (series.name if hasattr(series, 'name') else 'y')
    s = ensure_daily_series(series)

    # 2. detrend for FFT analysis
    trend, detrended = detrend_series(s)

    # 3. FFT main
    fft_main = compute_fft(detrended, dt=dt, windowed=True, max_period_plot=max_period_plot)
    candidates_main = find_fft_peaks(fft_main['periods'], fft_main['amps'], energy_thresh=fft_energy_thresh)

    # 4. FFT derivative
    dseries = s.diff().dropna()
    # detrend derivative as well
    _, d_detrended = detrend_series(dseries)
    fft_diff = compute_fft(d_detrended, dt=dt, windowed=True, max_period_plot=max_period_plot)
    candidates_diff = find_fft_peaks(fft_diff['periods'], fft_diff['amps'], energy_thresh=fft_energy_thresh)

    # 5. ACF/PACF
    acf_res = compute_acf_pacf(s, nlags=nlags_acf)
    acf_res_diff = compute_acf_pacf(dseries, nlags=nlags_acf)
    acf_lags = significant_acf_lags(acf_res['acf'], n=len(s.dropna()))
    acf_lags_diff = significant_acf_lags(acf_res_diff['acf'], n=len(dseries))

    # 6. classification
    classified = classify_fft_candidates(candidates_main, candidates_diff, acf_lags, acf_lags_diff,
                                         tol_abs=tolerance_abs, tol_rel=tolerance_rel,
                                         energy_thresh=fft_energy_thresh)

    # 7. candidate list for STL: take ones not noise/uncertain and include some business-relevant periods
    cand_periods = [c['period'] for c in classified if c['classification'] in ('real', 'uncertain')]
    # add common candidates
    cand_periods += [7, 14, 30]
    cand_periods = sorted(set([float(p) for p in cand_periods if p > 1]))

    # 8. STL scoring
    best_period, stl_scores = choose_best_period_by_stl(s, cand_periods)

    # 9. determine additive/multiplicative
    seasonality_type, diagnostics = determine_additive_or_multiplicative(s, best_period)

    # 10. prepare figures (FFT, FFT diff, ACF/PACF, STL)
    figs = {}

    # FFT plot
    fig1, ax1 = plt.subplots(figsize=(8,3))
    ax1.plot(fft_main['periods'], fft_main['amps'])
    ax1.scatter([p for p,a in candidates_main], [a for p,a in candidates_main], color='C1')
    ax1.set_xlim(0.5, max_period_plot)
    ax1.set_xlabel('Period (days)')
    ax1.set_ylabel('Amplitude')
    ax1.set_title(f'FFT ({sname})')
    figs['fft'] = fig1

    # FFT diff plot
    fig2, ax2 = plt.subplots(figsize=(8,3))
    ax2.plot(fft_diff['periods'], fft_diff['amps'])
    ax2.scatter([p for p,a in candidates_diff], [a for p,a in candidates_diff], color='C1')
    ax2.set_xlim(0.5, max_period_plot)
    ax2.set_xlabel('Period (days)')
    ax2.set_ylabel('Amplitude')
    ax2.set_title(f'FFT derivative ({sname})')
    figs['fft_diff'] = fig2

    # ACF/PACF plots
    fig3, ax3 = plt.subplots(2,1, figsize=(8,5))
    lags = np.arange(len(acf_res['acf']))
    ax3[0].stem(lags, acf_res['acf'])
    ax3[0].set_title('ACF (series)')
    ax3[0].set_xlim(0, min(nlags_acf, len(lags)-1))
    ax3[1].stem(np.arange(len(acf_res['pacf'])), acf_res['pacf'])
    ax3[1].set_title('PACF (series)')
    figs['acf_pacf'] = fig3

    fig4, ax4 = plt.subplots(2,1, figsize=(8,5))
    lagsd = np.arange(len(acf_res_diff['acf']))
    ax4[0].stem(lagsd, acf_res_diff['acf'])
    ax4[0].set_title('ACF (diff)')
    ax4[0].set_xlim(0, min(nlags_acf, len(lagsd)-1))
    ax4[1].stem(np.arange(len(acf_res_diff['pacf'])), acf_res_diff['pacf'])
    ax4[1].set_title('PACF (diff)')
    figs['acf_pacf_diff'] = fig4

    # STL decomposition with best_period
    if best_period >= 2:
        stl_res = STL(s.dropna(), period=best_period, robust=True).fit()
        fig5 = stl_res.plot()
        fig5.set_size_inches(8,6)
        figs['stl'] = fig5
    else:
        figs['stl'] = None
        stl_res = None

    # assemble output
    out = {
        'series_name': sname,
        'fft_main': {'periods': fft_main['periods'], 'amps': fft_main['amps'], 'candidates': candidates_main},
        'fft_diff': {'periods': fft_diff['periods'], 'amps': fft_diff['amps'], 'candidates': candidates_diff},
        'acf': {'vals': acf_res['acf'], 'pacf': acf_res['pacf'], 'significant_lags': acf_lags},
        'acf_diff': {'vals': acf_res_diff['acf'], 'pacf': acf_res_diff['pacf'], 'significant_lags': acf_lags_diff},
        'classified_candidates': classified,
        'candidate_periods_for_stl': cand_periods,
        'stl_scores': stl_scores,
        'best_period': best_period,
        'seasonality_type': seasonality_type,
        'seasonality_diagnostics': diagnostics,
        'figures': figs
    }

    return out


if __name__ == '__main__':
    result = analyze_series(df["VENTA"], col_name="VENTA")
    
    print("Periodo elegido:", result["best_period"])
    print("Tipo de estacionalidad:", result["seasonality_type"])

```

```python
# forecast_pipeline_adapted_v2.py
import os
import json
import numpy as np
import pandas as pd
from statsmodels.tsa.seasonal import STL
from xgboost import XGBRegressor
from arch import arch_model
from scipy.stats import t

# =====================================================
# 1. DATA PREPARATION
# =====================================================

def load_and_aggregate(D4NN, target_col):
    """
    Agrupa por FECHA_CORRIDA y conserva columnas originales (sin PORCENT_PROMO).
    """
    Frame = D4NN[['FECHA_OPERACION', target_col, 'MES', 'FOP']].copy()
    agg_dict = {col: 'mean' for col in Frame.columns if col not in ['FECHA_OPERACION', target_col]}
    agg_dict[target_col] = 'sum'

    df = Frame.groupby('FECHA_OPERACION').agg(agg_dict)
    df.index = pd.to_datetime(df.index)
    return df.sort_index()


def add_seasonal_and_calendar(df, target_col, period=14):
    """
    Descomposición con STL (robusto) — mantiene las columnas del primer programa.
    """
    stl = STL(df[target_col], period=period, robust=True)
    result = stl.fit()
    df['trend'] = result.trend

    df['DiaSemana'] = df.index.dayofweek

    df["sin_mes"] = np.sin(2 * np.pi * df["MES"] / 12)
    df["cos_mes"] = np.cos(2 * np.pi * df["MES"] / 12)
    df["sin_dia"] = np.sin(2 * np.pi * df["DiaSemana"] / 7)
    df["cos_dia"] = np.cos(2 * np.pi * df["DiaSemana"] / 7)

    return df, result


def add_business_features(df, json_path):
    """
    Lee JSON con DiaBueno / MesBueno y crea flags.
    """
    with open(json_path, 'r') as f:
        BC = json.load(f)
    df['DiaMayorBoletaje'] = df['DiaSemana'].isin(BC["DiaBueno"])
    df['MesMayorBoletaje'] = df['MES'].isin(BC["MesBueno"])
    return df, BC


def add_diff_features(df):
    """
    Mantiene exactamente las diferencias y lags del código que enviaste.
    """
    df['trend_diff'] = df['trend'].diff()
    df["diff_lag1"] = df["trend_diff"].shift(1)
    df["diff_lag7"] = df["trend_diff"].shift(7)
    df["diff_lag30"] = df["trend_diff"].shift(30)

    df["diff_ma7"] = df["trend_diff"].rolling(7).mean()
    df["diff_ma30"] = df["trend_diff"].rolling(30).mean()
    return df.dropna()


# =====================================================
# 2. TRAINING
# =====================================================

def build_train_test(df, feature_cols, target="trend_diff", test_ratio=0.2):
    X = df[feature_cols]
    y = df[target]
    split = int(len(df) * (1 - test_ratio))
    return X.iloc[:split], X.iloc[split:], y.iloc[:split], y.iloc[split:], split


def train_xgboost(X_train, y_train):
    """
    Hiperparámetros iguales a los del primer script que pegaste (800, lr=0.03, depth=6).
    """
    model = XGBRegressor(
        n_estimators=800,
        learning_rate=0.03,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        verbosity=0,
        objective="reg:squarederror"
    )
    model.fit(X_train, y_train)
    return model


def reconstruct_trend(df, y_pred, split):
    last = df['trend'].iloc[split - 1]
    trend_pred = np.cumsum(y_pred) + last
    trend_real = df["trend"].iloc[split:].values
    return trend_real, trend_pred


# =====================================================
# 3. FUTURE FORECAST
# =====================================================

def forecast_autoregressive(df, model, BC_json, days=15):
    """
    Forecast autoregresivo respetando el esquema original (sin PORCENT_PROMO).
    """
    last_date = df.index[-1]
    future = pd.DataFrame(index=pd.date_range(last_date + pd.Timedelta(1), periods=days))

    # calendar
    future["MES"] = future.index.month
    future["DiaSemana"] = future.index.dayofweek
    future["sin_mes"] = np.sin(2 * np.pi * future["MES"] / 12)
    future["cos_mes"] = np.cos(2 * np.pi * future["MES"] / 12)
    future["sin_dia"] = np.sin(2 * np.pi * future["DiaSemana"] / 7)
    future["cos_dia"] = np.cos(2 * np.pi * future["DiaSemana"] / 7)

    # business rules
    future["DiaMayorBoletaje"] = future["DiaSemana"].isin(BC_json["DiaBueno"])
    future["MesMayorBoletaje"] = future["MES"].isin(BC_json["MesBueno"])

    # exogenas (naive): solo FOP disponible
    future["FOP"] = df["FOP"].iloc[-1]

    # autoregressive loop
    diffs = []
    last_trend = df["trend"].iloc[-1]
    last_diff = df["trend_diff"].iloc[-1]

    for i in range(days):
        row = pd.DataFrame({
            "diff_lag1": [last_diff],
            "diff_lag7": [df["trend_diff"].iloc[-7 + i] if i < 7 else diffs[i - 7]],
            "diff_lag30": [df["trend_diff"].iloc[-30 + i] if i < 30 else diffs[i - 30]],
            "diff_ma7": [df["trend_diff"].iloc[-7:].mean()],
            "diff_ma30": [df["trend_diff"].iloc[-30:].mean()],
            "sin_dia": [future["sin_dia"].iloc[i]],
            "cos_dia": [future["cos_dia"].iloc[i]],
            "sin_mes": [future["sin_mes"].iloc[i]],
            "cos_mes": [future["cos_mes"].iloc[i]],
            "DiaMayorBoletaje": [future["DiaMayorBoletaje"].iloc[i]],
            "MesMayorBoletaje": [future["MesMayorBoletaje"].iloc[i]],
            "FOP": [future["FOP"].iloc[i]],
        })

        pred = model.predict(row)[0]
        diffs.append(pred)
        last_trend += pred
        last_diff = pred

    future["pred_trend"] = np.cumsum(diffs) + df["trend"].iloc[-1]
    return future


def apply_garch_intervals(resid, future_df, seasonal, alpha=0.99):
    """
    GARCH(1,1) sobre residuos y cálculo de intervalos — sigue la lógica original.
    """
    am = arch_model(resid, mean='Zero', vol='Garch', p=1, q=1, dist='t').fit(disp='off')
    var = am.forecast(horizon=len(future_df), reindex=False).variance.values[-1]
    std = np.sqrt(var)

    nu = am.params.get("nu", None)
    q = t.ppf(1 - (1 - alpha) / 2, nu) if nu else 1.96

    future_df["upper"] = future_df["pred_trend"] + q * std
    future_df["lower"] = future_df["pred_trend"] - q * std

    future_df["forecast_final"] = future_df["pred_trend"] + seasonal
    future_df["forecast_upper"] = future_df["upper"] + seasonal
    future_df["forecast_lower"] = future_df["lower"] + seasonal
    future_df["volatility"] = std
    return future_df


# =====================================================
# 4. ORQUESTACIÓN (Pipeline maestro)
# =====================================================

def run_forecast_pipeline(D4NN, target_col, json_path, future_days=15):
    df = load_and_aggregate(D4NN, target_col)
    df_final = df[[target_col]].copy()
    df_final[target_col + "_FORE"] = df[[target_col]].copy()
    df_final["FORE_LOWER"] = df[[target_col]].copy()
    df_final["FORE_UPPER"] = df[[target_col]].copy()

    df, result = add_seasonal_and_calendar(df, target_col)
    df, BC_json = add_business_features(df, json_path)
    df = add_diff_features(df)

    feature_cols = [
        "diff_lag1", "diff_lag7", "diff_lag30", "diff_ma7", "diff_ma30",
        "sin_dia", "cos_dia", "sin_mes", "cos_mes",
        "DiaMayorBoletaje", "MesMayorBoletaje", "FOP"
    ]

    # --- Train / predict ---
    Xtr, Xts, ytr, yts, split = build_train_test(df, feature_cols)
    model = train_xgboost(Xtr, ytr)

    pred = model.predict(Xts)
    real, recon = reconstruct_trend(df, pred, split)

    # --- Evaluación: RMSE, MAPE, R² sobre niveles del trend reconstruido ---
    from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error, r2_score
    rmse = np.sqrt(mean_squared_error(real, recon))
    mape = mean_absolute_percentage_error(real, recon) * 100
    r2 = r2_score(real, recon)

    print("\n📊 RESULTADOS XGBoost (Diferencias → Trend reconstruido) — EVALUACIÓN:")
    print(f"RMSE : {rmse:.4f}")
    print(f"MAPE : {mape:.2f}%")
    print(f"R²   : {r2:.6f}")

    resid = real - recon

    # --- Forecast futuro y GARCH ---
    fut = forecast_autoregressive(df, model, BC_json, future_days)
    seas = np.resize(result.seasonal[-14:].values, future_days)
    fut = apply_garch_intervals(resid, fut, seas)

    # Normalizar índices para evitar nanoseconds raros
    fut.index = fut.index.round("s")
    df_final.index = df_final.index.round("s")

    # Intersección segura (solo índices que existen en df_final)
    idx = fut.index.intersection(df_final.index)

    # Asignación sin riesgo de KeyError
    df_final.loc[idx, target_col + "_FORE"] = fut.loc[idx, "forecast_final"]
    df_final.loc[idx, "FORE_UPPER"] = fut.loc[idx, "forecast_upper"]
    df_final.loc[idx, "FORE_LOWER"] = fut.loc[idx, "forecast_lower"]

    df_final = df_final.fillna(0)

    return df_final, fut, model


# ================= EJECUCIÓN =================
if __name__ == "__main__":
    ruta_principal = os.getcwd()
    BuenasCarac_path = os.path.join(ruta_principal, "Files", "BuenaCaracteristicas.json")

    # Ejemplo de llamada (asegúrate de definir D4NN en tu entorno)
    df_forecast, future_df, model = run_forecast_pipeline(
         D4NN,
         target_col="VENTA",
         json_path=BuenasCarac_path,
         future_days=15
     )
```
