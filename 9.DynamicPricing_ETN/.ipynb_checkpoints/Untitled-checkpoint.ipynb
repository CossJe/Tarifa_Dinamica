# 1️⃣ Modelo con BOLETOS_VEND como target

```python
# ============================================================
#  XGBoost + GridSearch - Pronóstico de BOLETOS_VEND
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error, r2_score
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from datetime import timedelta

# =======================
# CONFIGURACIÓN
# =======================
forecast_horizon = 14  # días a pronosticar
cv_splits = 3          # particiones para validación cruzada temporal

# =======================
# CARGA DE DATOS
# =======================
df = pd.read_csv("Data4LSTM.csv", index_col="FECHA_CORRIDA")
df.index = pd.to_datetime(df.index)
df = df.sort_index()

if "BOLETOS_VEND_DIF" in df.columns:
    df = df.drop(columns=["BOLETOS_VEND_DIF"])

# =======================
# DIVISIÓN TRAIN / TEST TEMPORAL
# =======================
split_index = int(len(df) * 0.8)
train_df = df.iloc[:split_index]
test_df  = df.iloc[split_index:]

target_col = "BOLETOS_VEND"
X = df.drop(columns=[target_col])
y = df[target_col]

X_train, X_test = X.iloc[:split_index], X.iloc[split_index:]
y_train, y_test = y.iloc[:split_index], y.iloc[split_index:]

# =======================
# GRID SEARCH
# =======================
xgb = XGBRegressor(random_state=42)

param_grid = {
    "n_estimators": [200, 400, 600],
    "learning_rate": [0.01, 0.05, 0.1],
    "max_depth": [4, 6, 8],
    "subsample": [0.7, 0.8, 1.0],
    "colsample_bytree": [0.7, 0.8, 1.0]
}

tscv = TimeSeriesSplit(n_splits=cv_splits)

grid = GridSearchCV(
    estimator=xgb,
    param_grid=param_grid,
    cv=tscv,
    scoring="neg_root_mean_squared_error",
    verbose=2,
    n_jobs=-1
)

print("\n🔍 Buscando mejores parámetros...\n")
grid.fit(X_train, y_train)

print("\n✅ Mejores parámetros encontrados:")
print(grid.best_params_)
print(f"Mejor RMSE CV: {-grid.best_score_:.2f}")

# =======================
# MODELO FINAL CON LOS MEJORES PARÁMETROS
# =======================
best_model = grid.best_estimator_

best_model.fit(X_train, y_train)

# =======================
# PREDICCIÓN Y EVALUACIÓN
# =======================
y_pred = best_model.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mape = mean_absolute_percentage_error(y_test, y_pred) * 100
r2 = r2_score(y_test, y_pred)

print("\n===== MÉTRICAS TEST =====")
print(f"🔹 RMSE: {rmse:.2f}")
print(f"🔹 MAPE: {mape:.2f}%")
print(f"🔹 R²: {r2:.3f}")

# =======================
# VISUALIZACIÓN TEST
# =======================
plt.figure(figsize=(12, 5))
plt.plot(test_df.index, y_test, label="Real", linewidth=2)
plt.plot(test_df.index, y_pred, label="Predicho", linestyle="--")
plt.title("Predicción XGBoost - BOLETOS_VEND (Con GridSearch)")
plt.xlabel("Fecha")
plt.ylabel("Boletos vendidos")
plt.legend()
plt.show()
```

```python
# ================================================
# XGBoost para pronóstico de TENDENCIA
# ================================================
import pandas as pd
import numpy as np
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error, r2_score

df = df.copy()

# -------------------------
# Crear features de lags
# -------------------------
df["trend_lag1"] = df["trend"].shift(1)
df["trend_lag7"] = df["trend"].shift(7)
df["trend_lag30"] = df["trend"].shift(30)

# -------------------------
# Rolling windows
# -------------------------
df["trend_ma7"] = df["trend"].rolling(7).mean()
df["trend_ma30"] = df["trend"].rolling(30).mean()

# -------------------------
# Features de fecha
# -------------------------
df["sin_mes"] = np.sin(2 * np.pi * df["MES"] / 12)
df["cos_mes"] = np.cos(2 * np.pi * df["MES"] / 12)

# Eliminar filas con NaN generadas por lags y rolling
df = df.dropna()

# -------------------------
# Selección de features
# -------------------------
features = ["trend_lag1", "trend_lag7", "trend_lag30", "trend_ma7", "trend_ma30",
            "DiaSemana", "sin_mes", "cos_mes"]

X = df[features]
Y = df["trend"]

# dividir train/test respetando secuencia temporal (NO shuffle)
train_size = int(len(df) * 0.8)
X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_train, y_test = Y.iloc[:train_size], Y.iloc[train_size:]

# -------------------------
# Modelo XGBoost
# -------------------------
model = XGBRegressor(
    n_estimators=800,
    learning_rate=0.03,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror"
)

model.fit(X_train, y_train)

# -------------------------
# Predicciones y métricas
# -------------------------
y_pred = model.predict(X_test)

print("\n📊 RESULTADOS XGBoost (TENDENCIA):")
print("RMSE:", round(np.sqrt(mean_squared_error(y_test, y_pred)), 2))
print("MAPE:", round(mean_absolute_percentage_error(y_test, y_pred) * 100, 2), "%")
print("R²:", round(r2_score(y_test, y_pred), 4))
```
